```python
import matplotlib.pyplot as plt
import numpy as np
import random
from scipy.integrate import simps
```

```python
def BE(E,mu,t):
  return 1/(np.exp((E-mu)/(k*t))-1)

def FD(E,Ef,t):
  return 1/(np.exp((E-Ef)/(k*t))+1)
```

```python
k = 8.6*10**(-5)
mu,Ef = 0,1.0

T = [20,200,400]
E = np.linspace(0.01,2,1000)
```

```python
style = ['-','.-','--',':','-','.-']
st = iter(style)

for t in T:
  plt.plot(E,BE(E,mu,t),next(st),label="BE T = %4d "%t)
  plt.plot(E,FD(E,Ef,t),next(st),label="FD T = %4d "%t)

plt.ylabel("$BE,FD$ $unction$ ------>")
plt.xlabel("$Energ$ $in$ $eV$ ------>")
plt.legend(loc="best")
plt.axhline(0,linestyle='--',color='black')
plt.axvline(0,linestyle='--',color='black')
plt.title("$BE$ & $FD$ $Distribution.$")
plt.grid()

plt.show()
```
