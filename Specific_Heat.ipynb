```python
import matplotlib.pyplot as plt
import numpy as np
import random
from scipy.integrate import simps
```

```python
def f(x):
  return x**4*np.exp(x)/(np.exp(x)-1)**2

def MonteCarlo(a,b,N,f):
  x = np.random.uniform(a,b,N)
  return (b-a)*np.mean(f(x))
```

```python
N = 100000
Temp = np.linspace(1,50,1000)
Theta_D = 10
Theta_E = 10

Dulong_petit = np.ones(1000)

Debye = [3*(T/Theta_D)**3*MonteCarlo(0,Theta_D/T,N,f) for T in Temp]

Einstein = (Theta_E/Temp)**2*np.exp(Theta_E/Temp)/(np.exp(Theta_E/Temp)-1)**2
```

```python
plt.plot(Temp,Debye,label='Debye Plot.')
plt.plot(Temp,Dulong_petit,'-.',lw=2,label='Dulong Petit PLot.')
plt.plot(Temp,Einstein,'--',lw=1.5,label='Einstein Plot.')

plt.title("$The$ $Debye$ $Approximation$ & $Debye$ $Petit$.")
plt.xlabel("Temperature in K ------>")
plt.ylabel(r'$\frac{C_v}{3R}$ ----->')

plt.grid()
plt.legend(loc='best')
plt.axhline(0,linestyle='--',color='black')
plt.axvline(0,linestyle='--',color='black')

plt.show()
```
