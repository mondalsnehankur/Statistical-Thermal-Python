# ***Radioactive Decay***

```python
import numpy as np
from numpy import random
import matplotlib.pyplot as plt
```

```python
N = 1000
Nleft = N
lam = 0.01
t = 500
dt = 1
T = []
NLEFT = []

for t in range(0,t,dt):
  for i in range(Nleft):
    x = random.random()
    if(x<=lam):
      Nleft -= 1
  NLEFT.append(Nleft)
  T.append(t)
```

```python
# Plotting

plt.plot(T,NLEFT,label='Simulated')

tt = np.arange(0,500,0.01)
plt.plot(tt,1000*np.exp(-lam*tt),'--',label='Theoritical')

plt.title('Radioactive Decay.')
plt.xlabel('$Time$ ------->')
plt.ylabel('No:of remaining radionucleide.')
plt.legend(loc='best')
plt.grid()
plt.show()
```
