# ***Einstein's Specific Heat Theory.***

```python
import matplotlib.pyplot as plt
import numpy as np
import random
# from scipy.integrate import simps
```

```python
Temp = np.linspace(1,50,1000)
def Einstein(Theta_E,Temp):
  return(Theta_E/Temp)**2*np.exp(Theta_E/Temp)/(np.exp(Theta_E/Temp)-1)**2
```

```python
plt.plot(Temp,Einstein(20,Temp),label=r'$\Theta_E = 20 K$')
plt.plot(Temp,Einstein(10,Temp),'--',label=r'$\Theta_E = 10 K$')
plt.plot(Temp,Einstein(30,Temp),'.-',label=r'$\Theta_E = 30 K$')

plt.title("Speific-Heat Einstein's Theory.")
plt.legend(loc='best')
plt.xlabel("$Temperature$ ------>")
plt.ylabel(r'$\frac{C_v}{3R}$ ------>')
plt.grid()
plt.axhline(0,linestyle='--',color='black')
plt.axvline(0,linestyle='--',color='black')

plt.show()
```
