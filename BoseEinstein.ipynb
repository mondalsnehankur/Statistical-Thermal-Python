```python
import matplotlib.pyplot as plt
import numpy as np
import random
from scipy.integrate import simps
```

```python
def f(E,T):
  return 1.0/(np.exp((E-mu)/(K*T))-1)
```

```python
mu = 0.1  # in eV
K = 8.617*10**(-5)  # in eV/K

E = np.linspace(0.101,0.16,1000)  # in eV
T = [10,50,100,400]   # in K
```

```python
style =['-','.-','--',':']
st = iter(style)

for t in T:
  plt.title("$Bose$ $Einstein's$ $Distribution$ $for$ $various$ $Tmeperature.$")
  plt.plot(E,f(E,t),next(st),label="T = %8.2f"%(t))

plt.xlabel(r'$E$ ----->')
plt.ylabel("f(E,T) ------->")

plt.legend(loc="upper right")
plt.grid()
plt.axhline(0,linestyle='--',color='black')
plt.axvline(0.10,linestyle='--',color='black')

plt.xlim(0.09,0.17)
plt.show()
```
