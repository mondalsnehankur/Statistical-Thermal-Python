# ***Maxwell Boltzman Energy Distribution vs Temperature.***

```python
import matplotlib.pyplot as plt
import numpy as np
import random
from scipy.integrate import simps
```

```python
k =8.617*10**(-5)               # in eV per Kelvin
T = np.linspace(100,2000,1000)  # in eV
```

```python
def MBf(E,T):
  return(2.0/np.sqrt(np.pi))*(1.0/(k*T)**1.5)*E**0.5*np.exp(-E/(k*T))

E = [0.05,0.1,0.15]             # Temperature in Kelvin
A = []
```

```python
style =['-','.-','--']
st = iter(style)

for e in E:
  plt.plot(T,MBf(e,T),next(st),label="E =%8.2f K"%(e))

plt.title("Maxwell Boltzman Energy Distribution vs Temperature.")
plt.legend(loc='best')
plt.grid()
plt.xlabel("$Temperature$ $in$ $K$ ------->")
plt.ylabel("$MBf(e,T)$ ------>")
plt.axhline(0,linestyle='--',color='black')
plt.axvline(0,linestyle='--',color='black')

plt.show()
```
